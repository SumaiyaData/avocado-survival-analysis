```python
# ============================================================
# 1. LOAD LIBRARIES
# ============================================================

!pip install lifelines dowhy openpyxl ydata-profiling graphviz rpy2
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import math
import os
import zipfile
from lifelines import KaplanMeierFitter,CoxPHFitter,WeibullAFTFitter
from lifelines.statistics import multivariate_logrank_test, logrank_test
from lifelines.utils import restricted_mean_survival_time
from dowhy import CausalModel
from graphviz import Digraph


# ============================================================
# 2. EXTRACT AND LOAD DATA
# ============================================================

zip_path = ("/content/Hass Avocado Ripening Photographic Dataset.zip")

with zipfile.ZipFile(zip_path,"r") as z:
  z.extractall("/content/avocado")
avocado_original = pd.read_excel("/content/avocado/Hass Avocado Ripening Photographic Dataset/Avocado Ripening Dataset.xlsx")

avocado = avocado_original.copy()
avocado.head()


# ============================================================
# 3. CREATE FRUIT-DAY RIPENING DATASET
# ============================================================

photo_check = (avocado.groupby(["Storage Group","Sample","Day of Experiment"]).size())  # Checking expected two images per fruit per day
photo_check.value_counts()

duplicate_rows = avocado.duplicated().sum()  # Duplicate observation check
print("Duplicate rows:",duplicate_rows)
print("Total photographic observations:",len(avocado))

# ------------------------------------------------------------
# Primary aggregation:
# Maximum ripening stage from two fruit sides
# ------------------------------------------------------------


daily_ripening_state = (avocado.groupby(["Storage Group","Sample","Day of Experiment"])["Ripening Index Classification"].max().reset_index())
daily_ripening_state = daily_ripening_state.rename(columns={"Storage Group":"storage","Sample":"sample","Day of Experiment":"day","Ripening Index Classification":"stage"})
daily_ripening_state.insert(0,"record_id",range(1,len(daily_ripening_state)+1))
print("Number of fruit samples:",daily_ripening_state[["storage","sample"]].drop_duplicates().shape[0])
print(daily_ripening_state.groupby("storage")[["sample"]].nunique())


# ============================================================
# 4. CREATE SURVIVAL DATASET FUNCTION
# ============================================================



def create_shelf_life_dataset(daily_ripening_data,endpoint_stage):

    shelf_life_records = []

    fruit_groups = (daily_ripening_data.groupby(["storage","sample"]))

    for fruit_key, fruit_data in fruit_groups:

        storage, sample = fruit_key

        fruit_data = (fruit_data.sort_values("day"))


        endpoint_days = (fruit_data.loc[fruit_data["stage"] >= endpoint_stage,"day"])

        if len(endpoint_days) > 0:
            days = endpoint_days.iloc[0]
            event = 1
        else:
            days = fruit_data["day"].max()
            event = 0


        shelf_life_records.append([storage,sample,days,event])

    return pd.DataFrame(shelf_life_records,columns=["storage","sample","days","event"])


# ============================================================
# 5. CREATE PRIMARY SURVIVAL DATASET
# ============================================================


# Stage 4 = primary shelf-life endpoint


survival_df = create_shelf_life_dataset(daily_ripening_state,endpoint_stage=4)

print(survival_df.head())

print("Survival dataset shape:",survival_df.shape)


print(survival_df["event"].value_counts())

survival_df.to_csv("avocado_survival_dataset_stage4.csv",index=False)

print(survival_df.groupby("storage")["days"].describe())

print(pd.crosstab(survival_df["storage"],survival_df["event"]))



# ============================================================
# 6. SIDE AGGREGATION SENSITIVITY DATASETS
# ============================================================


daily_ripening_mean = (avocado.groupby(["Storage Group","Sample","Day of Experiment"])["Ripening Index Classification"].mean().reset_index())

daily_ripening_mean = daily_ripening_mean.rename(columns={"Storage Group":"storage","Sample":"sample","Day of Experiment":"day","Ripening Index Classification":"stage"})

daily_ripening_min = (avocado.groupby(["Storage Group","Sample","Day of Experiment"])["Ripening Index Classification"].min().reset_index())

daily_ripening_min = daily_ripening_min.rename(columns={"Storage Group":"storage","Sample":"sample","Day of Experiment":"day","Ripening Index Classification":"stage"})

survival_mean = create_shelf_life_dataset(daily_ripening_mean,endpoint_stage=4)

survival_min = create_shelf_life_dataset(daily_ripening_min,endpoint_stage=4)


# ============================================================
# 7. SIDE AGGREGATION SENSITIVITY COMPARISON
# ============================================================

sensitivity_datasets = {"Maximum side stage":survival_df,"Mean side stage":survival_mean, "Minimum side stage":survival_min}

for name, data in sensitivity_datasets.items():

    print("\n====================")
    print(name)
    print("====================")

    print(data.groupby("storage")["days"].median())

# KM sensitivity curves

for name, data in sensitivity_datasets.items():

    kmf = KaplanMeierFitter()

    plt.figure(figsize=(8,5))

    for group in data["storage"].unique():

        group_data = data[data["storage"] == group]

        kmf.fit(group_data["days"],group_data["event"],label=group)

        kmf.plot_survival_function()

    plt.title(f"KM Sensitivity: {name}")

    plt.xlabel("Days of Storage")

    plt.ylabel("Survival Probability")

    plt.grid(True)

    plt.show()


# ============================================================
# 8. CENSORED FRUIT INSPECTION
# ============================================================

censored_df = survival_df[survival_df["event"] == 0]
print("Number of censored fruits:",len(censored_df))

print(censored_df)
censored_df.to_csv("censored_fruits.csv",index=False)

censored_details = (daily_ripening_state.merge(censored_df[["storage","sample"]], on=["storage","sample"],how="inner").sort_values(["storage","sample","day"]).groupby(["storage","sample"]).tail(1))

print("Last observation details of censored fruits:")

print(censored_details)

censored_details.to_csv("censored_last_observation_details.csv",index=False)



# ============================================================
# 8A. CENSORING SENSITIVITY ANALYSIS
# Removing very short follow-up censored fruits
# ============================================================


short_censored = survival_df[(survival_df["event"] == 0)&(survival_df["days"] <= 1)]

print("Day-1-only censored fruits:",len(short_censored))

survival_no_short_censor = survival_df.drop(short_censored.index)

print("Original dataset:",survival_df.shape)

print("After removing short censoring:",survival_no_short_censor.shape)

print(survival_no_short_censor.groupby("storage")["days"].median())


kmf = KaplanMeierFitter() # KM after censoring sensitivity

plt.figure(figsize=(10,6))

for group in survival_no_short_censor["storage"].unique():

    data = survival_no_short_censor[survival_no_short_censor["storage"] == group]

    kmf.fit(durations=data["days"],event_observed=data["event"],label=group)

    kmf.plot_survival_function()

plt.title("Sensitivity Analysis: Excluding Day-1 Censored Fruits")

plt.xlabel("Days of Storage")

plt.ylabel("Survival Probability")

plt.grid(True)

plt.show()


# ============================================================
# 9. KAPLAN-MEIER SURVIVAL CURVE
# ============================================================

kmf = KaplanMeierFitter()

plt.figure(figsize=(10,6))

for group in survival_df["storage"].unique():

    data = survival_df[survival_df["storage"] == group]

    kmf.fit(durations=data["days"],event_observed=data["event"],label=group)

    kmf.plot_survival_function()


plt.title("Kaplan-Meier Curve: Avocado Shelf Life")
plt.xlabel("Days of Storage")
plt.ylabel("Probability of Remaining Before Stage 4")
plt.grid(True)
plt.savefig("kaplan_meier_curve.png",dpi=300,bbox_inches="tight")
plt.show()


# ============================================================
# 10. KM MEDIAN SURVIVAL TIME
# ============================================================

median_results = []

for group in survival_df["storage"].unique():

    data = survival_df[survival_df["storage"] == group]

    kmf.fit(durations=data["days"],event_observed=data["event"],label=group)

    median_results.append([group,kmf.median_survival_time_])

median_df = pd.DataFrame(median_results,columns=["Storage","Median_Survival_Days"])

print(median_df)

median_df.to_csv("KM_median_survival.csv",index=False)


# ============================================================
# 11. LOG-RANK TEST
# ============================================================

logrank_result = multivariate_logrank_test(survival_df["days"],survival_df["storage"],survival_df["event"])

print(logrank_result.summary)

logrank_result.summary.to_csv("logrank_results.csv")


# ============================================================
# 12. PAIRWISE LOG-RANK TEST
# T20 vs Tam
# ============================================================

t20_data = survival_df[survival_df["storage"] == "T20"]
tam_data = survival_df[survival_df["storage"] == "Tam"]
t20_tam_result = logrank_test(t20_data["days"],tam_data["days"],event_observed_A=t20_data["event"],event_observed_B=tam_data["event"])

print(t20_tam_result.summary)
t20_tam_result.summary.to_csv("logrank_T20_vs_Tam.csv")


# ============================================================
# 13. COX PROPORTIONAL HAZARDS MODEL
# ============================================================

cox_df = pd.get_dummies(survival_df,columns=["storage"],drop_first=False)
cox_df = cox_df.astype({"storage_T10":int,"storage_T20":int,"storage_Tam":int})
cox_df = cox_df.drop(columns=["storage_T10"])   # T10 reference group
cph = CoxPHFitter()
cph.fit(cox_df.drop(columns=["sample"]),duration_col="days",event_col="event")
cph.print_summary()
cox_results = cph.summary
cox_results.to_csv("cox_results.csv")
cph.check_assumptions(cox_df.drop(columns=["sample"]),p_value_threshold=0.05)

# ============================================================
# 14. WEIBULL ACCELERATED FAILURE TIME MODEL
# ============================================================

aft_df = pd.get_dummies(survival_df,columns=["storage"],drop_first=True)

aft = WeibullAFTFitter()

aft.fit(aft_df.drop(columns=["sample"]),duration_col="days",event_col="event")

aft.print_summary()
aft.summary.to_csv("aft_results.csv")


# ============================================================
# AUTOMATIC EXTRACTION OF AFT TIME RATIOS
# ============================================================

T20_ratio = aft.summary.loc[("lambda_", "storage_T20"),"exp(coef)"]
Tam_ratio = aft.summary.loc[("lambda_", "storage_Tam"),"exp(coef)"]
print("T20 time ratio:",T20_ratio)
print("Tam time ratio:",Tam_ratio)

# ============================================================
# 14A. INTERVAL-CENSORED WEIBULL AFT SENSITIVITY ANALYSIS
# ============================================================

# Purpose:
# The exact transition day to Stage 4 is unknown.
# It occurred between the last observed day before Stage 4
# and the first day when Stage 4 was detected.
#
# Example:
# Day 12 = Stage 3
# Day 13 = Stage 4
#
# True transition:
# 12 < event time <= 13
# ============================================================


interval_records = []


for (storage, sample), fruit_data in daily_ripening_state.groupby(["storage","sample"]):

    fruit_data = (fruit_data.sort_values("day"))
    endpoint = fruit_data[fruit_data["stage"] >= 4]

    if len(endpoint) > 0:

        upper = endpoint["day"].iloc[0]

        previous = fruit_data[fruit_data["day"] < upper]

        if len(previous) > 0:

            lower = previous["day"].iloc[-1]

        else:

            lower = 0

    else:

        lower = fruit_data["day"].max()          # Right censored observation

        upper = np.inf


    interval_records.append([storage,sample,lower,upper])

interval_df = pd.DataFrame(interval_records,columns=["storage", "sample","lower","upper"])

print(interval_df.head())



# ------------------------------------------------------------
# Run interval-censored Weibull AFT using R survival package
# ------------------------------------------------------------


%load_ext rpy2.ipython


import rpy2.robjects as robjects

from rpy2.robjects import pandas2ri


pandas2ri.activate()



robjects.globalenv["interval_df"] = interval_df



robjects.r(
'''

library(survival)


interval_df$storage <- as.factor(interval_df$storage)


interval_model <- survreg(

    Surv(
        lower,
        upper,
        type="interval2"
    )

    ~ storage,

    data=interval_df,

    dist="weibull"

)


print(summary(interval_model))


capture.output(
    summary(interval_model),
    file="interval_censored_weibull_AFT_results.txt"
)

'''
)





# ============================================================
# 15. DoWhy Analysis
# T20 vs T10
# ============================================================


causal_T20 = survival_df[survival_df["storage"].isin(["T10","T20"])].copy()

causal_T20 = pd.get_dummies(causal_T20,columns=["storage"])

causal_T20["storage_T20"] = (causal_T20["storage_T20"].astype(int))

causal_T20_data = causal_T20[["days","storage_T20"]]

model_T20 = CausalModel(data=causal_T20_data,treatment="storage_T20",outcome="days",graph="""digraph {storage_T20 -> days;}""")

estimand_T20 = model_T20.identify_effect()

estimate_T20 = model_T20.estimate_effect(estimand_T20,method_name="backdoor.linear_regression")

print("Observed T20 vs T10 difference:",estimate_T20.value)

print(model_T20.refute_estimate(estimand_T20,estimate_T20,method_name="random_common_cause"))

print(model_T20.refute_estimate(estimand_T20,estimate_T20,method_name="placebo_treatment_refuter"))

print(model_T20.refute_estimate(estimand_T20,estimate_T20,method_name="data_subset_refuter"))


# ============================================================
# 16. DoWhy Analysis
# Tam vs T10
# ============================================================


causal_Tam = survival_df[survival_df["storage"].isin(["T10","Tam"])].copy()
causal_Tam = pd.get_dummies(causal_Tam,columns=["storage"])
causal_Tam["storage_Tam"] = (causal_Tam["storage_Tam"].astype(int))
causal_Tam_data = causal_Tam[["days","storage_Tam"]]

model_Tam = CausalModel(data=causal_Tam_data,treatment="storage_Tam",outcome="days",graph="""digraph {storage_Tam -> days;}""")

estimand_Tam = model_Tam.identify_effect()

estimate_Tam = model_Tam.estimate_effect(estimand_Tam,method_name="backdoor.linear_regression")

print("Observed Tam vs T10 difference:",estimate_Tam.value)

print(model_Tam.refute_estimate(estimand_Tam,estimate_Tam,method_name="random_common_cause"))

print(model_Tam.refute_estimate(estimand_Tam,estimate_Tam,method_name="placebo_treatment_refuter"))

print(model_Tam.refute_estimate(estimand_Tam,estimate_Tam,method_name="data_subset_refuter"))


# ============================================================
# 17. E-VALUE SENSITIVITY
# ============================================================


def calculate_evalue(hr):

    return hr + np.sqrt(hr * (hr - 1))

HR_T20 = np.exp(cph.params_["storage_T20"])

HR_Tam = np.exp(cph.params_["storage_Tam"])

print("Hazard Ratio T20:",HR_T20)

print("Hazard Ratio Tam:",HR_Tam)

print("E-value T20:",calculate_evalue(HR_T20))

print("E-value Tam:",calculate_evalue(HR_Tam))


# Note:
# E-values are retained as supplementary sensitivity outputs.
# They are NOT headline findings because:
# 1. Cox HRs are inflated due to limited overlap.
# 2. Assignment mechanism is not fully documented.





# ============================================================
# 18. CONCEPTUAL DAG WITH STORAGE CHAMBER
# ============================================================


dag = Digraph("Avocado_Conceptual_DAG",format="png")

dag.attr(rankdir="LR")

# Nodes

dag.node("U","Baseline traits\n(maturity,size,batch)")

dag.node("C","Storage chamber")

dag.node("S","Storage condition\n(T10,T20,Tam)")

dag.node("M","Ripening progression")

dag.node("Y","Shelf-life\n(days to Stage 4)")


# Baseline effects

dag.edge("U","M")

dag.edge("U","Y")

# Chamber effects

dag.edge("C","S")

dag.edge("C","M")

dag.edge("C","Y")

# Storage effects

dag.edge("S","M")

dag.edge("M","Y")

dag.render("avocado_conceptual_DAG",view=True)


# ============================================================
# 19. STAGE 5 SENSITIVITY ANALYSIS
# ============================================================


survival_stage5 = create_shelf_life_dataset(daily_ripening_state,endpoint_stage=5)
print("Stage 5 dataset:",survival_stage5.shape)
print(survival_stage5["event"].value_counts())
print(survival_stage5.groupby("storage")["days"].describe())

kmf = KaplanMeierFitter()  # Stage 5 KM Curve
plt.figure(figsize=(10,6))

for group in survival_stage5["storage"].unique():
    data = survival_stage5[survival_stage5["storage"] == group]

    kmf.fit(data["days"],data["event"],label=group)
    kmf.plot_survival_function()

plt.title("Sensitivity Analysis: Stage 5 Endpoint")
plt.xlabel("Days of Storage")
plt.ylabel("Probability Remaining Before Stage 5")
plt.grid(True)
plt.legend(title="Storage Group")
plt.savefig("KM_stage5_sensitivity.png",dpi=300,bbox_inches="tight")
plt.show()

stage5_cox = pd.get_dummies(survival_stage5,columns=["storage"],drop_first=False) # Stage 5 Cox model
stage5_cox = stage5_cox.astype({"storage_T10":int,"storage_T20":int,"storage_Tam":int})
stage5_cox = stage5_cox.drop(columns=["storage_T10"])
cph_stage5 = CoxPHFitter()
cph_stage5.fit(stage5_cox.drop(columns=["sample"]),duration_col="days",event_col="event")
cph_stage5.print_summary()
cph_stage5.summary.to_csv("cox_stage5_results.csv")


# ============================================================
# 20. RESTRICTED MEAN SURVIVAL TIME (RMST)
# ============================================================


tau = survival_df["days"].max()

def calculate_rmst(data):

    kmf = KaplanMeierFitter()

    kmf.fit(durations=data["days"],event_observed=data["event"])

    return restricted_mean_survival_time(kmf,t=tau)

rmst_results = []

for group in survival_df["storage"].unique():

    data = survival_df[survival_df["storage"] == group]

    rmst_results.append([group,calculate_rmst(data)])

rmst_df = pd.DataFrame(rmst_results,columns=["Storage","RMST_days"])

print(rmst_df)

rmst_df.to_csv("RMST_results.csv",index=False)


# ============================================================
# 20A. RMST BOOTSTRAP CONFIDENCE INTERVALS
# ============================================================

def bootstrap_rmst(data,n_boot=1000):

    values = []

    for i in range(n_boot):
      boot = data.sample(frac=1,replace=True)
      values.append(calculate_rmst(boot))

    return (np.percentile(values,2.5),np.percentile(values,97.5))

rmst_ci_results = []

for group in survival_df["storage"].unique():
  data = survival_df[survival_df["storage"] == group]

  lower, upper = bootstrap_rmst(data)

  rmst_ci_results.append([group,calculate_rmst(data),lower,upper])

rmst_final_df = pd.DataFrame(rmst_ci_results, columns=["Storage","RMST_days","RMST_lower95","RMST_upper95"])

print(rmst_final_df)
rmst_final_df.to_csv("RMST_with_bootstrap_CI.csv",index=False)


# ============================================================
# 21. SHELF-LIFE DISTRIBUTION
# ============================================================

plt.figure(figsize=(8,5))
sns.boxplot(data=survival_df,x="storage",y="days")
plt.title("Shelf-life Distribution by Storage Condition")
plt.xlabel("Storage Condition")
plt.ylabel("Days to Stage 4")
plt.savefig("shelf_life_distribution.png", dpi=300,bbox_inches="tight")
plt.show()


# ============================================================
# 22. FINAL RESULTS SUMMARY TABLE
# ============================================================

rmst_reporting = rmst_df.merge(rmst_final_df[["Storage","RMST_lower95","RMST_upper95"]],on="Storage")

final_results = pd.DataFrame({

"Analysis":["Kaplan-Meier","Log-rank test","KM median survival","Weibull AFT","Interval-censored Weibull AFT",
"RMST with bootstrap CI","Cox PH","DoWhy T20 vs T10","DoWhy Tam vs T10"], "Main Finding":[
"T10 maintained the highest survival probability", f"p-value = {logrank_result.p_value:.3e}",
str(median_df.to_dict(orient="records")), f"T20 retained {T20_ratio:.2f} and Tam retained {Tam_ratio:.2f} of T10 shelf-life",
"Interval-censored model confirmed consistent storage effects", str(rmst_reporting.to_dict(orient="records")),
"Strong storage differences detected; hazard ratios interpreted cautiously due to limited overlap",
f"Observed difference = {estimate_T20.value:.2f} days", f"Observed difference = {estimate_Tam.value:.2f} days"]})


print(final_results)
final_results.to_csv("final_analysis_summary_revised.csv",index=False)
print("The analysis has been completed successfully.")

```
