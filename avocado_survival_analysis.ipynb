```python
# ============================================================
# 1. LOAD LIBRARIES
# ============================================================

!pip install lifelines openpyxl graphviz rpy2
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import zipfile
from lifelines import KaplanMeierFitter,CoxPHFitter,WeibullAFTFitter,LogNormalAFTFitter, LogLogisticAFTFitter
from lifelines.statistics import multivariate_logrank_test, logrank_test
from lifelines.utils import restricted_mean_survival_time,median_survival_times
from graphviz import Digraph


# ============================================================
# 2. EXTRACT AND LOAD DATA
# ============================================================

zip_path = ("/content/Hass Avocado Ripening Photographic Dataset.zip")

with zipfile.ZipFile(zip_path,"r") as z:
  z.extractall("/content/avocado")

avocado_original = pd.read_excel("/content/avocado/Hass Avocado Ripening Photographic Dataset/Avocado Ripening Dataset.xlsx")
avocado = avocado_original.copy()
avocado.head()


# ============================================================
# 3. CREATE FRUIT-DAY RIPENING DATASET
# ============================================================

photo_check = (avocado.groupby(["Storage Group","Sample","Day of Experiment"]).size())  # Checking expected two images per fruit per day
print(photo_check.value_counts())

duplicate_rows = avocado.duplicated().sum()  # Duplicate observation check
print("Duplicate rows:",duplicate_rows)
print("Total photographic observations:",len(avocado))

daily_ripening_state = (avocado.groupby(["Storage Group","Sample","Day of Experiment"])["Ripening Index Classification"].max().reset_index())
daily_ripening_state = daily_ripening_state.rename(columns={"Storage Group":"storage","Sample":"sample","Day of Experiment":"day","Ripening Index Classification":"stage"})
daily_ripening_state.insert(0,"record_id",range(1,len(daily_ripening_state)+1))
print("Number of fruit samples:",daily_ripening_state[["storage","sample"]].drop_duplicates().shape[0])
print(daily_ripening_state.groupby("storage")[["sample"]].nunique())


# ============================================================
# 4. CREATE SURVIVAL DATASET FUNCTION
# ============================================================

def create_shelf_life_dataset(daily_ripening_data,endpoint_stage):

    shelf_life_records = []

    fruit_groups = (daily_ripening_data.groupby(["storage","sample"]))

    for fruit_key, fruit_data in fruit_groups:

         storage, sample = fruit_key

         fruit_data = (fruit_data.sort_values("day"))


         endpoint_days = (fruit_data.loc[fruit_data["stage"] >= endpoint_stage,"day"])

         if len(endpoint_days) > 0:
             days = endpoint_days.iloc[0]
             event = 1
         else:
             days = fruit_data["day"].max()
             event = 0

         shelf_life_records.append([storage,sample,days,event])

    return pd.DataFrame(shelf_life_records,columns=["storage","sample","days","event"])


# ============================================================
# 5. CREATE PRIMARY SURVIVAL DATASET
# ============================================================

side_check = (avocado.groupby(["Storage Group","Sample","Day of Experiment"])["Ripening Index Classification"].agg(["min","max"]).reset_index())

side_check["side_difference"] = (side_check["max"] - side_check["min"])

print("Fruit-days with different side ratings:",(side_check["side_difference"] > 0).sum())

print("Total fruit-days:",len(side_check))

survival_df = create_shelf_life_dataset(daily_ripening_state,endpoint_stage=4)

print(survival_df.head())
print("Survival dataset shape:",survival_df.shape)
print(survival_df["event"].value_counts())
survival_df.to_csv("avocado_survival_dataset_stage4.csv",index=False)
print(survival_df.groupby("storage")["days"].describe())
print(pd.crosstab(survival_df["storage"],survival_df["event"]))


# ============================================================
# 6. CENSORED FRUIT INSPECTION
# ============================================================

censored_df = survival_df[survival_df["event"] == 0]
print("Number of censored fruits:",len(censored_df))

print(censored_df)
censored_df.to_csv("censored_fruits.csv",index=False)

censored_details = (daily_ripening_state.merge(censored_df[["storage","sample"]], on=["storage","sample"],how="inner").sort_values(["storage","sample","day"]).groupby(["storage","sample"]).tail(1))

print("Last observation details of censored fruits:")

print(censored_details)

censored_details.to_csv("censored_last_observation_details.csv",index=False)


# ============================================================
# 7. CENSORING SENSITIVITY ANALYSIS
# ============================================================

short_censored = survival_df[(survival_df["event"] == 0)&(survival_df["days"] <= 1)]

print("Day-1-only censored fruits:",len(short_censored))

survival_no_short_censor = survival_df.drop(short_censored.index)

print("Original dataset:",survival_df.shape)

print("After removing short censoring:",survival_no_short_censor.shape)

print(survival_no_short_censor.groupby("storage")["days"].median())

kmf = KaplanMeierFitter() # KM after censoring sensitivity

plt.figure(figsize=(10,6))

for group in survival_no_short_censor["storage"].unique():

    data = survival_no_short_censor[survival_no_short_censor["storage"] == group]

    kmf.fit(durations=data["days"],event_observed=data["event"],label=group)

    kmf.plot_survival_function()

plt.title("Sensitivity Analysis: Excluding Day-1 Censored Fruits")

plt.xlabel("Days of Storage")

plt.ylabel("Survival Probability")

plt.grid(True)

plt.show()


# ============================================================
# 8. EXTREME CENSORING SENSITIVITY SCENARIOS
# ============================================================

latest_followup = survival_df.copy()

group_end_days = (daily_ripening_state.groupby("storage")["day"].max().to_dict())

for group, end_day in group_end_days.items():
  mask = ((latest_followup["storage"] == group) & (latest_followup["event"] == 0))
  latest_followup.loc[mask, "days"] = end_day

latest_followup.loc[latest_followup["event"] == 0,"event"] = 0

earliest_event = survival_df.copy()

earliest_event.loc[earliest_event["event"] == 0,"days"] += 1

earliest_event.loc[earliest_event["event"] == 0,"event"] = 1

print("Latest event bound:")
print(latest_followup.groupby("storage")["days"].median())
print("\nEarliest event bound:")
print(earliest_event.groupby("storage")["days"].median())


# ============================================================
# 9. AFT ANALYSIS UNDER EXTREME CENSORING SCENARIOS
# ============================================================

def run_weibull_aft(data):

    temp = pd.get_dummies(data,columns=["storage"],drop_first=True)
    model = WeibullAFTFitter()
    model.fit(temp.drop(columns=["sample"]),duration_col="days",event_col="event")
    T20_ratio = model.summary.loc[("lambda_", "storage_T20"),"exp(coef)"]
    Tam_ratio = model.summary.loc[("lambda_", "storage_Tam"),"exp(coef)"]
    return T20_ratio, Tam_ratio

latest_T20, latest_Tam = run_weibull_aft(latest_followup)
earliest_T20, earliest_Tam = run_weibull_aft(earliest_event)

print("LATEST-BOUND TIME RATIOS")
print("T20:", latest_T20)
print("Tam:", latest_Tam)

print("\nEARLIEST-BOUND TIME RATIOS")
print("T20:", earliest_T20)
print("Tam:", earliest_Tam)


# ============================================================
# 10. KAPLAN-MEIER SURVIVAL CURVE
# ============================================================

kmf = KaplanMeierFitter()

plt.figure(figsize=(10,6))

for group in survival_df["storage"].unique():

    data = survival_df[survival_df["storage"] == group]

    kmf.fit(durations=data["days"],event_observed=data["event"],label=group)

    kmf.plot_survival_function()


plt.title("Kaplan-Meier Curve: Avocado Shelf Life")
plt.xlabel("Days of Storage")
plt.ylabel("Probability of Remaining Before Stage 4")
plt.grid(True)
plt.savefig("kaplan_meier_curve.png",dpi=300,bbox_inches="tight")
plt.show()


# ============================================================
# 11. KM MEDIAN SURVIVAL TIME WITH 95% CI
# ============================================================

median_results = []

for group in survival_df["storage"].unique():

    data = survival_df[survival_df["storage"] == group]

    kmf = KaplanMeierFitter()

    kmf.fit(durations=data["days"],event_observed=data["event"],label=group)

    median = kmf.median_survival_time_

    median_ci = median_survival_times(kmf.confidence_interval_)

    lower = median_ci.filter(like="lower").iloc[0, 0]

    upper = median_ci.filter(like="upper").iloc[0, 0]

    median_results.append([group,median,lower,upper])

median_df = pd.DataFrame(median_results,columns=["Storage","Median_Survival_Days","Median_CI_Lower","Median_CI_Upper"])

print("\nKM MEDIANS WITH 95% CI")
print(median_df)
median_df.to_csv("KM_median_survival_with_CI.csv",index=False)


# ============================================================
# 12. LOG-RANK TEST
# ============================================================

logrank_result = multivariate_logrank_test(survival_df["days"],survival_df["storage"],survival_df["event"])
print(logrank_result.summary)
logrank_result.summary.to_csv("logrank_results.csv")


# ============================================================
# 13. PAIRWISE LOG-RANK TEST (T20 vs Tam)
# ============================================================

t20_data = survival_df[survival_df["storage"] == "T20"]
tam_data = survival_df[survival_df["storage"] == "Tam"]
t20_tam_result = logrank_test(t20_data["days"],tam_data["days"],event_observed_A=t20_data["event"],event_observed_B=tam_data["event"])

print(t20_tam_result.summary)
t20_tam_result.summary.to_csv("logrank_T20_vs_Tam.csv")


# ============================================================
# 14. COX PROPORTIONAL HAZARDS MODEL
# ============================================================

cox_df = pd.get_dummies(survival_df,columns=["storage"],drop_first=False)
cox_df = cox_df.astype({"storage_T10":int,"storage_T20":int,"storage_Tam":int})
cox_df = cox_df.drop(columns=["storage_T10"])   # T10 reference group
cph = CoxPHFitter()
cph.fit(cox_df.drop(columns=["sample"]),duration_col="days",event_col="event")
cph.print_summary()
cox_results = cph.summary
cox_results.to_csv("cox_results.csv")
cph.check_assumptions(cox_df.drop(columns=["sample"]),p_value_threshold=0.05)


# ============================================================
# 15. WEIBULL ACCELERATED FAILURE TIME MODEL
# ============================================================

aft_df = pd.get_dummies(survival_df,columns=["storage"],drop_first=True)
aft = WeibullAFTFitter()

after_fit = aft.fit(aft_df.drop(columns=["sample"]),duration_col="days",event_col="event")

aft.print_summary()
aft.summary.to_csv("aft_results.csv")


# ============================================================
# 16. AFT DISTRIBUTIONAL SENSITIVITY (Weibull vs Log-normal vs Log-logistic)
# ============================================================

aft_sensitivity_df = pd.get_dummies(survival_df,columns=["storage"],drop_first=True)

aft_input = aft_sensitivity_df.drop(columns=["sample"])

# Weibull
weibull_sens = WeibullAFTFitter()
weibull_sens.fit(aft_input,duration_col="days",event_col="event")

# Log-normal
lognormal_sens = LogNormalAFTFitter()
lognormal_sens.fit(aft_input,duration_col="days",event_col="event")

# Log-logistic
loglogistic_sens = LogLogisticAFTFitter()
loglogistic_sens.fit(aft_input,duration_col="days",event_col="event")

aft_distribution_comparison = pd.DataFrame({
    "Model": ["Weibull","Log-normal","Log-logistic"],
    "AIC": [weibull_sens.AIC_,lognormal_sens.AIC_,loglogistic_sens.AIC_],
    "T20_vs_T10_TR": [weibull_sens.summary.loc[("lambda_", "storage_T20"), "exp(coef)"],
        np.exp(lognormal_sens.summary.loc[("mu_", "storage_T20"), "coef"]),loglogistic_sens.summary.loc[("alpha_", "storage_T20"), "exp(coef)"]],
    "Tam_vs_T10_TR": [weibull_sens.summary.loc[("lambda_", "storage_Tam"), "exp(coef)"],
        np.exp(lognormal_sens.summary.loc[("mu_", "storage_Tam"), "coef"]),
        loglogistic_sens.summary.loc[("alpha_", "storage_Tam"), "exp(coef)"]]})

aft_distribution_comparison = (aft_distribution_comparison.sort_values("AIC"))

print("\nAFT DISTRIBUTIONAL SENSITIVITY")
print(aft_distribution_comparison)

aft_distribution_comparison.to_csv("AFT_distribution_sensitivity.csv",index=False)

T20_ratio = aft.summary.loc[("lambda_", "storage_T20"),"exp(coef)"]
Tam_ratio = aft.summary.loc[("lambda_", "storage_Tam"),"exp(coef)"]
print("T20 time ratio:",T20_ratio)
print("Tam time ratio:",Tam_ratio)


# ============================================================
# 17. PRIMARY INTERVAL-CENSORED WEIBULL AFT ANALYSIS
# ============================================================

interval_records = []

for (storage, sample), fruit_data in daily_ripening_state.groupby(["storage","sample"]):

    fruit_data = (fruit_data.sort_values("day"))
    endpoint = fruit_data[fruit_data["stage"] >= 4]

    if len(endpoint) > 0:

        upper = endpoint["day"].iloc[0]

        previous = fruit_data[fruit_data["day"] < upper]

        if len(previous) > 0:

            lower = previous["day"].iloc[-1]

        else:

            lower = 0

    else:

        lower = fruit_data["day"].max()          # Right censored observation

        upper = np.inf


    interval_records.append([storage,sample,lower,upper])

interval_df = pd.DataFrame(interval_records,columns=["storage", "sample","lower","upper"])

print(interval_df.head())


# ============================================================
# 18. Run interval-censored Weibull AFT using R survival package
# ============================================================

%load_ext rpy2.ipython

import rpy2.robjects as robjects

from rpy2.robjects import pandas2ri

pandas2ri.activate()

robjects.globalenv["interval_df"] = interval_df

# FIT INTERVAL-CENSORED WEIBULL AFT MODEL IN R AND EXTRACT RESULTS

robjects.r(
'''

library(survival)

cat("\nR version:", R.version.string, "\n")
cat("survival package version:", as.character(packageVersion("survival")),"\n")

# Confirm interval-censoring structure

interval_df$storage <- relevel(as.factor(interval_df$storage),ref = "T10")
cat("Number of interval-data rows:", nrow(interval_df), "\\n")
cat("Number of right-censored rows:",sum(is.infinite(interval_df$upper)),"\\n")

# Convert infinite upper bounds to NA for explicit
# right-censoring representation in Surv(type="interval2")

interval_df$upper[is.infinite(interval_df$upper)] <- NA_real_

cat("Right-censored rows represented as NA:",sum(is.na(interval_df$upper)),"\n")

# Fit Weibull interval-censored AFT model

interval_model <- survreg(

    Surv(
        lower,
        upper,
        type="interval2"
    ) ~ storage,

    data = interval_df,dist = "weibull")


print(summary(interval_model))

# Extract coefficients and standard errors

coef_table <- summary(interval_model)$table

# Storage coefficients
beta_T20 <- coef(interval_model)["storageT20"]
beta_Tam <- coef(interval_model)["storageTam"]


# Standard errors
se_T20 <- coef_table["storageT20", "Std. Error"]
se_Tam <- coef_table["storageTam", "Std. Error"]


# Convert coefficients to AFT time ratios

TR_T20 <- exp(beta_T20)
TR_Tam <- exp(beta_Tam)

# 95% confidence intervals
T20_lower <- exp(beta_T20 - 1.96 * se_T20)
T20_upper <- exp(beta_T20 + 1.96 * se_T20)

Tam_lower <- exp(beta_Tam - 1.96 * se_Tam)
Tam_upper <- exp(beta_Tam + 1.96 * se_Tam)


# Create clean results table

interval_results <- data.frame(

    Comparison = c("T20 vs T10","Tam vs T10"),

    Coefficient = c(beta_T20,beta_Tam),

    SE = c(se_T20,se_Tam),

    Time_Ratio = c(TR_T20,TR_Tam),

    CI_Lower_95 = c(T20_lower,Tam_lower),

    CI_Upper_95 = c(T20_upper,Tam_upper))

cat("\\nINTERVAL-CENSORED WEIBULL AFT RESULTS\\n")

print(interval_results)

write.csv(interval_results,"interval_censored_weibull_AFT_results.csv",row.names = FALSE)

capture.output(summary(interval_model),interval_results,file = "interval_censored_weibull_AFT_results.txt")

''')


# ============================================================
# 19. IMPORT INTERVAL-CENSORED AFT RESULTS INTO PYTHON
# ============================================================

interval_results_py = pd.read_csv("interval_censored_weibull_AFT_results.csv")

print("\nInterval-censored Weibull AFT results imported into Python:")
print(interval_results_py)

interval_T20 = interval_results_py.loc[interval_results_py["Comparison"] == "T20 vs T10"].iloc[0]
interval_Tam = interval_results_py.loc[interval_results_py["Comparison"] == "Tam vs T10"].iloc[0]

interval_T20_ratio = interval_T20["Time_Ratio"]
interval_T20_low = interval_T20["CI_Lower_95"]
interval_T20_high = interval_T20["CI_Upper_95"]

interval_Tam_ratio = interval_Tam["Time_Ratio"]
interval_Tam_low = interval_Tam["CI_Lower_95"]
interval_Tam_high = interval_Tam["CI_Upper_95"]

print(f"T20 vs T10: TR={interval_T20_ratio:.3f} " f"(95% CI {interval_T20_low:.3f}-{interval_T20_high:.3f})")

print(f"Tam vs T10: TR={interval_Tam_ratio:.3f} " f"(95% CI {interval_Tam_low:.3f}-{interval_Tam_high:.3f})")


# ============================================================
# 20. STANDARD VS INTERVAL-CENSORED AFT COMPARISON
# ============================================================

aft_model_comparison = pd.DataFrame({

    "Model": [
        "Standard Weibull AFT",
        "Standard Weibull AFT",
        "Interval-censored Weibull AFT",
        "Interval-censored Weibull AFT"
    ],

    "Comparison": [
        "T20 vs T10",
        "Tam vs T10",
        "T20 vs T10",
        "Tam vs T10"
    ],

    "Time_Ratio": [
        T20_ratio,
        Tam_ratio,
        interval_T20_ratio,
        interval_Tam_ratio
    ],

    "CI_Lower_95": [
        aft.summary.loc[
            ("lambda_", "storage_T20"),
            "exp(coef) lower 95%"
        ],

        aft.summary.loc[
            ("lambda_", "storage_Tam"),
            "exp(coef) lower 95%"
        ],

        interval_T20_low,
        interval_Tam_low],

    "CI_Upper_95": [
        aft.summary.loc[
            ("lambda_", "storage_T20"),
            "exp(coef) upper 95%"],

        aft.summary.loc[
            ("lambda_", "storage_Tam"),
            "exp(coef) upper 95%"], interval_T20_high,interval_Tam_high]})


print("\nStandard versus interval-censored AFT comparison:")
print(aft_model_comparison)

aft_model_comparison.to_csv("AFT_standard_vs_interval_comparison.csv",index=False)


# ============================================================
# 22. CONCEPTUAL DAG WITH STORAGE CHAMBER
# ============================================================


dag = Digraph("Avocado_Conceptual_DAG",format="png")

dag.attr(rankdir="LR")

# Nodes

dag.node("U","Baseline traits\n(maturity,size,batch)")

dag.node("C","Storage chamber")

dag.node("S","Storage condition\n(T10,T20,Tam)")

dag.node("M","Ripening progression")

dag.node("Y","Shelf-life\n(days to Stage 4)")

# Baseline effects

dag.edge("U","M")

dag.edge("U","Y")

# Chamber effects

dag.edge("C","S")

dag.edge("C","M")

dag.edge("C","Y")

# Storage effects

dag.edge("S","M")

dag.edge("M","Y")

dag.render("avocado_conceptual_DAG",view=True)


# ============================================================
# 23. STAGE 5 SENSITIVITY ANALYSIS
# ============================================================

survival_stage5 = create_shelf_life_dataset(daily_ripening_state,endpoint_stage=5)

# STAGE 5: T20 VS TAM LOG-RANK TEST

stage5_t20 = survival_stage5[survival_stage5["storage"] == "T20"]
stage5_tam = survival_stage5[survival_stage5["storage"] == "Tam"]
stage5_t20_tam_result = logrank_test(stage5_t20["days"],stage5_tam["days"],event_observed_A=stage5_t20["event"],event_observed_B=stage5_tam["event"])

print("\nStage 5 T20 vs Tam log-rank test")
print(stage5_t20_tam_result.summary)
stage5_t20_tam_result.summary.to_csv("stage5_T20_vs_Tam_logrank.csv")

# STAGE 5 KM MEDIANS

stage5_medians = []

for group in survival_stage5["storage"].unique():

    data = survival_stage5[survival_stage5["storage"] == group]

    kmf.fit(data["days"],data["event"])

    stage5_medians.append([group,kmf.median_survival_time_])

stage5_median_df = pd.DataFrame(stage5_medians,columns=["Storage","Stage5_median_days"])

print(stage5_median_df)


# STAGE 5 WEIBULL AFT TIME RATIOS

stage5_aft_df = pd.get_dummies(survival_stage5,columns=["storage"],drop_first=True)

stage5_aft = WeibullAFTFitter()

stage5_aft.fit(stage5_aft_df.drop(columns=["sample"]),duration_col="days",event_col="event")

stage5_T20_ratio = stage5_aft.summary.loc[("lambda_", "storage_T20"),"exp(coef)"]

stage5_Tam_ratio = stage5_aft.summary.loc[("lambda_", "storage_Tam"),"exp(coef)"]

print("Stage 5 T20 time ratio:", stage5_T20_ratio)
print("Stage 5 Tam time ratio:", stage5_Tam_ratio)

stage5_aft.summary.to_csv("stage5_weibull_AFT_results.csv")
print("Stage 5 dataset:",survival_stage5.shape)
print(survival_stage5["event"].value_counts())
print(survival_stage5.groupby("storage")["days"].describe())

kmf = KaplanMeierFitter()  # Stage 5 KM Curve
plt.figure(figsize=(10,6))

for group in survival_stage5["storage"].unique():
    data = survival_stage5[survival_stage5["storage"] == group]
    kmf.fit(data["days"],data["event"],label=group)
    kmf.plot_survival_function()

plt.title("Sensitivity Analysis: Stage 5 Endpoint")
plt.xlabel("Days of Storage")
plt.ylabel("Probability Remaining Before Stage 5")
plt.grid(True)
plt.legend(title="Storage Group")
plt.savefig("KM_stage5_sensitivity.png",dpi=300,bbox_inches="tight")
plt.show()

stage5_cox = pd.get_dummies(survival_stage5,columns=["storage"],drop_first=False) # Stage 5 Cox model
stage5_cox = stage5_cox.astype({"storage_T10":int,"storage_T20":int,"storage_Tam":int})
stage5_cox = stage5_cox.drop(columns=["storage_T10"])
cph_stage5 = CoxPHFitter()
cph_stage5.fit(stage5_cox.drop(columns=["sample"]),duration_col="days",event_col="event")
cph_stage5.print_summary()
cph_stage5.summary.to_csv("cox_stage5_results.csv")


# ============================================================
# 24. RESTRICTED MEAN SURVIVAL TIME (RMST)
# ============================================================


tau = survival_df["days"].max()

print("RMST restriction horizon:", tau)
print("Bootstrap samples:", 1000)

def calculate_rmst(data):

    kmf = KaplanMeierFitter()

    kmf.fit(durations=data["days"],event_observed=data["event"])

    return restricted_mean_survival_time(kmf,t=tau)

rmst_results = []

for group in survival_df["storage"].unique():

    data = survival_df[survival_df["storage"] == group]

    rmst_results.append([group,calculate_rmst(data)])

rmst_df = pd.DataFrame(rmst_results,columns=["Storage","RMST_days"])

print(rmst_df)

rmst_df.to_csv("RMST_results.csv",index=False)


# ============================================================
# 25. RMST BOOTSTRAP CONFIDENCE INTERVALS
# ============================================================

def bootstrap_rmst(data, n_boot=1000, seed=42):

    rng = np.random.default_rng(seed)

    values = []

    for _ in range(n_boot):

         indices = rng.integers(0,len(data),len(data))

         boot = data.iloc[indices]
         values.append(calculate_rmst(boot))

    return (np.percentile(values, 2.5),np.percentile(values, 97.5))


# RMST estimates + bootstrap 95% CI for each storage group

rmst_ci_results = []

for i, group in enumerate(survival_df["storage"].unique()):

    data = survival_df[survival_df["storage"] == group]

    lower, upper = bootstrap_rmst(data, n_boot=1000,seed=42 + i)

    rmst_ci_results.append([group,calculate_rmst(data),lower,upper])


rmst_final_df = pd.DataFrame(rmst_ci_results,columns=["Storage","RMST_days","RMST_lower95","RMST_upper95"])

print("\nRMST WITH BOOTSTRAP 95% CI")
print(rmst_final_df)
rmst_final_df.to_csv("RMST_with_bootstrap_CI.csv",index=False)


# ============================================================
# 26. PAIRWISE RMST DIFFERENCES
# ============================================================

rmst_dict = dict(zip(rmst_final_df["Storage"],rmst_final_df["RMST_days"]))
print("\nPAIRWISE RMST POINT DIFFERENCES")
print("T10 - T20:",rmst_dict["T10"] - rmst_dict["T20"])
print("T10 - Tam:",rmst_dict["T10"] - rmst_dict["Tam"])
print("T20 - Tam:",rmst_dict["T20"] - rmst_dict["Tam"])


# ============================================================
# 27. BOOTSTRAP 95% CI FOR PAIRWISE RMST DIFFERENCES
# ============================================================

def bootstrap_rmst_difference(data_a,data_b,n_boot=1000,seed=42):

    rng = np.random.default_rng(seed)

    differences = []

    for _ in range(n_boot):

         idx_a = rng.integers(0,len(data_a),len(data_a))

         idx_b = rng.integers(0,len(data_b),len(data_b))

         boot_a = data_a.iloc[idx_a]
         boot_b = data_b.iloc[idx_b]

         diff = (calculate_rmst(boot_a) - calculate_rmst(boot_b))

         differences.append(diff)

    return (np.percentile(differences, 2.5), np.percentile(differences, 97.5))


comparisons = [("T10", "T20"),("T10", "Tam"),("T20", "Tam")]

rmst_pairwise_results = []

for i, (group_a, group_b) in enumerate(comparisons):

    data_a = survival_df[survival_df["storage"] == group_a]

    data_b = survival_df[survival_df["storage"] == group_b]

    point_difference = (calculate_rmst(data_a) - calculate_rmst(data_b))

    lower, upper = bootstrap_rmst_difference(data_a,data_b, n_boot=1000, seed=100 + i)

    rmst_pairwise_results.append([f"{group_a} - {group_b}",point_difference,lower,upper])

rmst_difference_df = pd.DataFrame(rmst_pairwise_results,columns=["Comparison","RMST_Difference","CI_Lower_95","CI_Upper_95"])

print("\nPAIRWISE RMST DIFFERENCES WITH 95% CI")
print(rmst_difference_df)

rmst_difference_df.to_csv("RMST_pairwise_differences_with_CI.csv",index=False)


# ============================================================
# 28. SHELF-LIFE DISTRIBUTION
# ============================================================

plt.figure(figsize=(8,5))
sns.boxplot(data=survival_df,x="storage",y="days")
plt.title("Observed Follow-up Time by Storage Condition")
plt.xlabel("Storage Condition")
plt.ylabel("Observed or Censoring Time (days)")
plt.savefig("shelf_life_distribution.png", dpi=300,bbox_inches="tight")
plt.show()


# ============================================================
# 29. FINAL RESULTS SUMMARY TABLE
# Main manuscript-oriented results
# ============================================================

rmst_reporting = rmst_df.merge(rmst_final_df[["Storage", "RMST_lower95", "RMST_upper95"]],on="Storage")

final_results = pd.DataFrame({

    "Analysis": [

         "Kaplan-Meier",

         "KM median survival",

         "Global log-rank",

         "T20 vs Tam log-rank",

         "Standard Weibull AFT",

         "Interval-censored Weibull AFT (primary)",

         "RMST with bootstrap CI",

         "RMST pairwise differences",

         "Censoring-bound sensitivity",

         "Stage 5 sensitivity"],

    "Main Finding": [

         "T10 showed the longest time remaining below Stage 4",

         str(
             median_df.to_dict(
                 orient="records")),

         f"p = {logrank_result.p_value:.3e}",

         f"p = {t20_tam_result.p_value:.3f}",

         (
             f"T20 TR={T20_ratio:.3f}; "
             f"Tam TR={Tam_ratio:.3f}"
         ),

         (
             f"T20 vs T10: TR={interval_T20_ratio:.3f} "
             f"(95% CI {interval_T20_low:.3f}-"
             f"{interval_T20_high:.3f}); "
             f"Tam vs T10: TR={interval_Tam_ratio:.3f} "
             f"(95% CI {interval_Tam_low:.3f}-"
             f"{interval_Tam_high:.3f})"
         ),

         str(
             rmst_reporting.to_dict(
                 orient="records"
             )
         ),

         str(
             rmst_difference_df.to_dict(
                 orient="records"
             )
         ),

         (
             f"Latest-bound TRs: "
             f"T20={latest_T20:.3f}, "
             f"Tam={latest_Tam:.3f}; "
             f"Earliest-bound TRs: "
             f"T20={earliest_T20:.3f}, "
             f"Tam={earliest_Tam:.3f}"
         ),

         (
             f"Stage 5 medians: "
             f"{stage5_median_df.to_dict(orient='records')}; "
             f"T20 TR={stage5_T20_ratio:.3f}; "
             f"Tam TR={stage5_Tam_ratio:.3f}")]})


print(final_results)

final_results.to_csv("final_analysis_summary_revised.csv",index=False)
print("The analysis has been completed successfully.")


# ============================================================
# 30. SOFTWARE VERSIONS
# ============================================================

import sys
import lifelines
import pandas
import numpy
import rpy2

print("\nSOFTWARE VERSIONS")
print("Python:", sys.version)
print("lifelines:", lifelines.__version__)
print("pandas:", pandas.__version__)
print("numpy:", numpy.__version__)
print("rpy2:", rpy2.__version__)
```
